Método para concatenar dos cadenas

```python
def concatenar(cadena1, cadena2):
    """
    Esta función concatena dos cadenas y devuelve el resultado.

    Args:
        cadena1 (str): La primera cadena a concatenar.
        cadena2 (str): La segunda cadena a concatenar.

    Returns:
        str: La cadena resultante de la concatenación de cadena1 y cadena2.
    """
    resultado = cadena1 + cadena2
    return resultado

if __name__ == "__main__":
    # Llamada al método concatenar()
    cadena = concatenar("Hola", " Mundo")
    print(cadena)
```

Método para comparar si un número es par o impar

```python
def es_par(numero):
    """
    Esta función verifica si un número es par o impar y devuelve el resultado.

    Args:
        numero (int): El número a verificar.

    Returns:
        bool: True si el número es par, False si el número es impar.
    """
    return numero % 2 == 0


if __name__ == "__main__":
    # Llamada al método es_par()
    numero = 5
    if es_par(numero):
        print(f"{numero} es par")
    else:
        print(f"{numero} es impar")
```

Método para obtener un número aleatorio del 0 al 100

```python
# Libreria random para obtener el número aleatorio
import random

def aleatorio():
    """
    Esta función devuelve un número aleatorio entre 0 y 100.

    Returns:
        int: Un número aleatorio entre 0 y 100.
    """
    numero = random.randint(0, 100)
    return numero


if __name__ == "__main__":
    # Llamada al método aleatorio()
    numero = aleatorio()
    print(numero)
```

Clase que representa un perro y método para simular un ladrido

```python
class Perro:
    """
    Esta clase representa un perro y tiene un método para simular un ladrido.
    """
    def ladrar(self):
        """
        Este método simula un ladrido.

        Returns:
            str: El sonido del ladrido.
        """
        return "guau"


if __name__ == "__main__":
    # Llamada al método ladrar()
    perro = Perro()
    print(perro.ladrar())
```

Clase para simular una cuenta bancaria, utilizando métodos de clase

```python
class CuentaBancaria:
    """Clase que representa una cuenta bancaria con funcionalidades básicas de depósito y verificación de estado."""

    def __init__(self, titular: str, saldo: float = 0):
        """Inicializa una nueva cuenta bancaria con un titular y un saldo inicial de 0.

        Args:
            titular (str): Nombre del titular de la cuenta.
            saldo (float): Saldo inicial de la cuenta.
        """
        self.titular = titular
        self._saldo = saldo

    @property
    def saldo(self) -> float:
        """Getter para el saldo de la cuenta.

        Returns:
            float: Saldo actual de la cuenta.
        """
        return self._saldo

    @saldo.setter
    def saldo(self, nuevo_saldo: float):
        """Setter para el saldo de la cuenta.

        Args:
            nuevo_saldo (float): Nuevo saldo a establecer.
        """
        self._saldo = nuevo_saldo

    @saldo.deleter
    def saldo(self):
        """Deleter para el saldo de la cuenta.

        Raises:
            AttributeError: No se puede eliminar el saldo directamente.
        """
        raise AttributeError("No se puede eliminar el saldo directamente.")

    @property
    def estado(self) -> str:
        """Retorna el estado de la cuenta basado en el saldo.

        Returns:
            str: "Cuenta activa" si el saldo es mayor o igual a 0, "Cuenta en descubierto" si el saldo es negativo.
        """
        if self.saldo >= 0:
            estado = "Cuenta activa"
        else:
            estado = "Cuenta en descubierto"
        return estado

    def depositar(self, monto: float) -> str:
        """Realiza un depósito en la cuenta bancaria si el monto es válido.

        Args:
            monto (float): Monto a depositar.

        Returns:
            str: Mensaje indicando el resultado del depósito.
        """
        if self.es_monto_valido(monto):
            self.saldo += monto
            resultado = f"Depósito exitoso. Nuevo saldo: {self.saldo}"
        else:
            resultado = "Error: Monto inválido. No se puede realizar el depósito."
        return resultado

    @classmethod
    def cuenta_nueva(cls, titular: str) -> "CuentaBancaria":
        """Crea una nueva cuenta bancaria con un saldo inicial de 0.

        Args:
            titular (str): Nombre del titular de la cuenta.

        Returns:
            CuentaBancaria: Una nueva instancia de CuentaBancaria con saldo inicial de 0.
        """
        return cls(titular, 0)

    @staticmethod
    def es_monto_valido(monto: float) -> bool:
        """Verifica si el monto es válido (mayor que cero).

        Args:
            monto (float): Monto a verificar.

        Returns:
            bool: True si el monto es mayor que cero, False en caso contrario.
        """
        return monto > 0


if __name__ == "__main__":
    # Llamada a la función de clase para crear una nueva cuenta bancaria
    cuenta = CuentaBancaria.cuenta_nueva("Victor")

    # Prueba de la función de depósito
    print(cuenta.depositar(100))  # Depósito exitoso. Nuevo saldo: 100

    # Prueba de la función de depósito con un monto inválido
    print(cuenta.depositar(-50))  # Error: Monto inválido. No se puede realizar el depósito.

    # Prueba de la propiedad estado
    print(cuenta.estado)  # Cuenta activa

    # Prueba del getter de saldo
    print(cuenta.saldo)          # 100 — accede directo, sin paréntesis

    # Prueba del setter de saldo
    cuenta.saldo = 500
    print(cuenta.saldo)          # 500 — el setter lo actualizó

    # Prueba del deleter de saldo (debe fallar a propósito)
    try:
        del cuenta.saldo
    except AttributeError as e:
        print(f"Error: {e}")
```
